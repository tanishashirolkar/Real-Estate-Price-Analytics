## 2. Data Cleaning and Preprocessing


```python
#remove the extra column
# Drop Unnamed column
df.drop('Unnamed: 0', axis=1, inplace=True)

# Check updated dataset
df.head()
```

### Missing Data Treatment ###

```python
# Fill missing bedroom values
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].mean())

# Fill missing bathroom values
df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].mean())
```


## 3. Exploratory Data Analysis (EDA)


```python
import matplotlib.pyplot as plt
import seaborn as sns
```


### EDA Q1: Distribution of House Prices


```python
# Distribution of house prices
plt.hist(df['price'], bins=30)

plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# Statistical summary of house prices
df['price'].describe()
```

```python
# Calculate skewness
df['price'].skew()
```


**Observation:**
- The histogram shows a right-skewed distribution.
- The mean house price is approximately $540,088, while the median is $450,000.
- House prices range from $75,000 to $7,700,000.
- The calculated skewness is approximately 4.02, indicating strong positive skewness.

**Conclusion:**

House prices are strongly right-skewed, with a small number of high-priced properties extending the distribution toward larger values. The median provides a useful complementary measure of a typical house price because it is less sensitive to extreme values than the mean.

**Further analysis:**

A logarithmic transformation can be explored to reduce skewness for visualization or future statistical modelling. The original house prices will be retained for the initial analysis.


### EDA Q2: Relationship Between Living Area and House Price


```python
# Relationship between living area and price
plt.scatter(df['sqft_living'], df['price'])

plt.title('Living Area vs Price')
plt.xlabel('Living Area (sq ft)')
plt.ylabel('Price')
plt.show()

```



**Observation:**
- The scatter plot shows a positive relationship between living area and house price.
- Houses with larger living areas generally tend to have higher sale prices.
- Considerable variation exists among houses with similar living areas.
- A few properties have unusually large living areas or high sale prices.

**Conclusion:**

Living area is an important variable associated with house prices. However, other property characteristics and location may also contribute to price differences. Further analysis is required to understand these relationships.


### EDA Q3: Relationship Between Bedrooms and House Price


```python
# Relationship between bedrooms and price
df_plot = df[
    (df['bedrooms'] % 1 == 0) &
    (df['bedrooms'] >= 1) &
    (df['bedrooms'] <= 10)
]

plt.figure(figsize=(12, 5))

sns.boxplot(x='bedrooms', y='price', data=df_plot)

plt.title('Bedrooms vs Price')
plt.xlabel('Bedrooms')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
print(df.columns.tolist())
```

```python
df['bedrooms'].value_counts().sort_index()
```


**Observation:**
- Median house prices generally tend to increase across several bedroom categories.
- House prices vary considerably within the same bedroom category.
- Multiple high-price outliers are visible.
- The analysis focuses on houses with 1–10 whole-number bedrooms; unusual bedroom counts are excluded from this visualization and may require separate investigation.

**Conclusion:**

Bedroom count is associated with differences in house-price distributions, but it does not explain prices on its own. Other property characteristics should also be considered when analysing house prices.

**Data-quality note:** The graph includes only houses with whole-number bedroom counts from 1 to 10. The original bedrooms column remains unchanged.


### EDA Q4: Relationship Between Bathrooms and House Price


```python
# Relationship between bathrooms and house price
sns.scatterplot(x='bathrooms', y='price', data=df)

plt.title('Bathrooms vs House Price')
plt.xlabel('Number of Bathrooms')
plt.ylabel('House Price ($)')
plt.show()
```

### EDA Q6: Correlation Between House Price and Property Features


```python
# Correlation among key variables
corr = df[['price', 'sqft_living', 'bedrooms',
           'bathrooms', 'grade']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```


**Observation:**
- Living area has the strongest positive correlation with price among the selected variables (0.70).
- Grade also has a strong positive correlation with price (0.67).
- Bathrooms have a moderate positive correlation with price (0.53).
- Bedrooms have a comparatively weaker positive correlation with price (0.31).
- Living area is positively correlated with both bathrooms and grade (0.76 for each).

**Conclusion:**

Among the selected variables, living area has the strongest linear association with house price, followed by grade. These relationships suggest that property size and grade are important factors to investigate further. Correlation does not establish causation, and other variables may also influence house prices.

## EDA Q7: Price per Square Foot Across Locations

**Objective:** To calculate the price per square foot for each property and compare typical values across ZIP codes to understand differences in housing costs.


```python
# Check the living area column
df['sqft_living'].describe()
```

```python
# Check invalid or missing living area values
df['sqft_living'].isnull().sum(), (df['sqft_living'] <= 0).sum()
```

```python
# Calculate price per square foot
df['price_per_sqft'] = df['price'] / df['sqft_living']

# Display the first five rows
df[['price', 'sqft_living', 'price_per_sqft']].head()
```

```python
# Calculate median price per square foot for each ZIP code
df.groupby('zipcode')['price_per_sqft'].median().sort_values(ascending=False)
```

```python
# Display the top 10 ZIP codes by median price per square foot
df.groupby('zipcode')['price_per_sqft'].median().sort_values(ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt

# Calculate the top 10 ZIP codes
top_10 = df.groupby('zipcode')['price_per_sqft'].median().sort_values(ascending=False).head(10)

# Create a bar chart
top_10.plot(kind='bar', figsize=(10, 5))

# Add title and labels
plt.title('Top 10 ZIP Codes by Median Price per Square Foot')
plt.xlabel('ZIP Code')
plt.ylabel('Median Price per Square Foot ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation –

- ZIP code **98039** has the highest median price per square foot among the 10 locations shown, at approximately **$565.17**.

- ZIP code **98103** has the lowest median price per square foot among these top 10 locations, at approximately **$360.05**.
- The chart shows that the median price per square foot varies across ZIP codes.
- Location may be an important factor when comparing housing costs.


**Conclusion:** ZIP code 98039 has the highest median price per square foot among the top 10 ZIP codes in this dataset. However, price per square foot alone does not determine whether a property offers better value.

- Comparing the price chart with the median living area and grade table suggests that property characteristics and location are both useful factors to investigate.

**Conclusion:** ZIP code 98039 has the highest median house price among the 10 ZIP codes analyzed. These results describe the recorded sales in this dataset and do not necessarily reflect current market prices.

## EDA Q9: House Sale Price Trends Over Time

**Objective:** To analyze how typical house sale prices changed over time using the sale dates recorded in the dataset.

```python
# Check the data type of the date column
df['date'].dtype
```

```python
# Check the earliest and latest sale dates
df['date'].min(), df['date'].max()
```

```python
# Calculate monthly median house prices
monthly_price = df.groupby(df['date'].dt.to_period('M'))['price'].median()

# Display the monthly median prices
monthly_price
```

```python
import matplotlib.pyplot as plt

# Create a line graph of monthly median house prices
monthly_price.plot(kind='line', marker='o', figsize=(10, 5))

# Add title and axis labels
plt.title('Monthly Median House Prices (2014–2015)')
plt.xlabel('Month')
plt.ylabel('Median House Price ($)')

# Display the graph
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation – EDA Q9: House Sale Price Trends Over Time

- The monthly median house price reached its highest value of $476,500 in April 2015.

- The lowest monthly median price was $425,545 in February 2015.
- Median prices declined during several months from mid-2014 to early 2015, although the trend was not continuous.
- Prices increased from February to April 2015 before decreasing in May 2015.

**Conclusion:** The monthly median sale prices fluctuated during the period covered by the dataset. The highest median was recorded in April 2015, while the lowest was recorded in February 2015. These findings describe historical sales in the dataset and should not be interpreted as current market trends.
